# Variables in Python
 
Variables are fundamental building blocks in programming that are used to store data values. These values can be referenced, modified, and manipulated throughout a program.
 
In Python, variables are created automatically when a value is assigned to them. Unlike some programming languages, Python does not require explicit variable declaration or data type specification before use.

# Why Do We Need Variables?
 
Imagine a student management system where you need to store:
 
- Student Name
- Student Age
- Student Marks
- Student Email
 
Instead of repeatedly typing the values, we store them in variables and use them whenever required.
 
### Real World Example
 
Without Variables:
 
```python
print("Avinash")
print("Avinash")
print("Avinash")
```
 
With Variables:
 
```python
student_name = "Avinash"
 
print(student_name)
print(student_name)
print(student_name)
```
 
Output:
 
```python
Avinash
Avinash
Avinash
```
 
Variables make programs easier to maintain and update.
 
---
 
# Creating Variables
 
Variables are created when a value is assigned.
 
### Syntax
 
```python
variable_name = value
```
 
### Example
 
```python
name = "Avinash"
age = 28
salary = 50000
```
 
---
 
# Multiple Variable Assignments
 
### Assign Different Values
 
```python
name = "Avinash"
age = 28
city = "Mumbai"
```
 
### Assign Multiple Values in One Line
 
```python
name, age, city = "Avinash", 28, "Mumbai"
 
print(name)
print(age)
print(city)
```
 
---
 
### Assign Same Value to Multiple Variables
 
```python
x = y = z = 100
 
print(x)
print(y)
print(z)
```
 
Output:
 
```python
100
100
100
```
 
---

# Checking Variable Type
 
Use the `type()` function.
 
### Example
 
```python
age = 25
 
print(type(age))
```
 
Output:
 
```python
<class 'int'>
```
 
---
 
### Example
 
```python
name = "Python"
 
print(type(name))
```
 
Output:
 
```python
<class 'str'>
```
 
---
 
# Dynamic Typing in Python
 
Python is dynamically typed.
 
A variable can change its data type during execution.
 
### Example
 
```python
data = 100
 
print(type(data))
```
 
Output:
 
```python
<class 'int'>
```
 
Now assign a string:
 
```python
data = "Python"
 
print(type(data))
```
 
Output:
 
```python
<class 'str'>
```
 
The same variable stores different types of values.

# Type Conversion
 
Sometimes we need to convert one data type to another.
 
## Convert String to Integer
 
```python
age = "25"
 
age = int(age)
 
print(type(age))
```
 
Output:
 
```python
<class 'int'>
```
 
---
 
## Convert Integer to String
 
```python
num = 100
 
num = str(num)
 
print(type(num))
```
 
Output:
 
```python
<class 'str'>
```
 
---
 
## Convert Integer to Float
 
```python
marks = 90
 
marks = float(marks)
 
print(type(marks))
```
 
Output:
 
```python
<class 'float'>
```

# Common Errors
 
## Using Variable Before Assignment
 
```python
print(name)
```
 
Output:
 
```python
NameError: name 'name' is not defined
```
 
Correct:
 
```python
name = "Python"
 
print(name)
```
 
---
 
## Invalid Variable Name
 
```python
2marks = 90
```
 
Output:
 
```python
SyntaxError
```
 
Correct:
 
```python
marks2 = 90
```
 
---
 
## Case Sensitivity
 
```python
name = "Python"
 
Name = "Programming"
 
print(name)
print(Name)
```
 
Output:
 
```python
Python
Programming
```
 
Python treats them as different variables.

# Key Points
 
- Variables are used to store data.
- Variables are created automatically when assigned.
- Python is dynamically typed.
- Variable names should be meaningful.
- Variable names cannot start with numbers.
- Variable names cannot contain spaces.
- Python is case-sensitive.
- Use `type()` to check data types.
- Type conversion can be performed using `int()`, `float()`, and `str()` functions.
- Variables improve code readability, maintainability, and reusability.
