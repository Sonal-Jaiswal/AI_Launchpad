```python
emp_id = input("Enter Employee ID: ")
emp_name = input("Enter your name: ")
current_performance = int(input("Enter your current performance score (out of 10): "))
salary = int(input("Enter your monthly salary: "))

if current_performance > 0 and current_performance <= 10:

    if current_performance >= 8:
        rating = 1
        percent = 15
        salary1 = salary * 1.15

    elif current_performance >= 6:
        rating = 2
        percent = 7
        salary1 = salary * 1.07

    elif current_performance >= 3:
        rating = 3
        percent = 5
        salary1 = salary * 1.05

    else:
        rating = 0
        percent = 0

    print("\n-------------------------------------------")
    print(f"Emp ID: {emp_id}")
    print(f"Name: {emp_name}")
    print(f"Performance Score: {current_performance}/10")
    print(f"Rating Received: {rating}")
    print(f"Percent Hike: {percent}%")
    print(f"Your Salary was: ₹ {salary:.2f} INR")
    print(f"Your expected Salary: ₹ {salary1:.2f} INR")
    print(f"Monthly increase: ₹ {salary1 - salary:.2f} INR") 
    print(f"Annual Salary: ₹ {salary1 * 12:.2f} INR") 
    print(f"Annual increase: ₹ {(salary1 - salary) * 12:.2f} INR") 
    print("-------------------------------------------")

else:
    print("Provide correct performance score out of 10 only.")
```

```python
Account_Number =12345678
PIN = 1234
balance = 10000

Account_Num = int(input("Enter Account Number: "))
if Account_Number == Account_Num:
    PIN_ = int(input("Enter PIN: "))

    if PIN == PIN_:
        print("Access Granted")
        print("1. Deposit Money")
        print("2. Withdraw Money")
        print("3. Check Balance")
        print("--------------------------------------")
        choice = int(input("Enter Choice: "))
        if choice == 1:
            deposit = float(input("Enter Deposit Amount: "))
            if deposit > 0:
                balance = balance + deposit
                print("Deposit Successful")
                print("Available Balance =", balance)
            else:
                print("Invalid Amount")
        elif choice == 2:
            withdraw = float(input("Enter Withdrawal Amount: "))

            if withdraw <= 0:
                print("Invalid Amount")
            elif withdraw > 25000:
                print("Transaction Limit Exceeded")
            elif withdraw > balance:
                print("Insufficient Balance")
            else:
                balance = balance - withdraw
                print("Withdrawal Successful")
                print("Available Balance =", balance)
        elif choice == 3:
            print("Current Balance =", balance)
        else:
            print("Invalid Menu Choice")

    else:
        print("Invalid PIN")

else:
    print("Invalid Account Number")

print("--------------------------------------")
```
